```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv('tip.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

## Univariate Analysis

```python
sns.countplot(x='size',data=df)
plt.show()
```

```python
sns.countplot(data=df,x=df['day'])
```

```python
sns.countplot(data=df,x=df['time'])
```

# Observation

### 1.The size of 2 members booking is greater compared to others
### 2.The size of 1 and 6 members have equal percentage also have least number
### 3.The customer bookings are most in saturday and least in friday 
### 4.There are no customer in first 3 days of the week
### 5.There are more customers during dinner compared to lunch
