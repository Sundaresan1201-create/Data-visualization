```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[2,4,6,8,10]
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.xlim(1,10)
plt.ylim(5,15)
plt.title("Line plot example")
plt.plot(x,y)
plt.savefig("Sample.png")
plt.show
```

```python
x=[111,112,113,114,115]
y=[40,95,35,96,38]
x1=[111,112,113,114,115]
y1=[95,40,96,38,98]
plt.xlabel("Roll-no")
plt.ylabel("Marks")
plt.xlim(110,116)
plt.ylim(1,100)
plt.title("Students performance")
plt.plot(x,y,label='line1',color="blue",linestyle="dotted",linewidth=3,marker='*',markerfacecolor='red',markersize=10)
plt.plot(x1,y1,label='line2',color="violet",linestyle="dashed",linewidth=3,marker='*',markerfacecolor='blue',markersize=10)
plt.legend()
plt.show
```
